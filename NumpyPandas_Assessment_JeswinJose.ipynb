```python
# 11
df=pd.read_csv("/content/drive/MyDrive/Data Science/Assessmment/numpy_pandas_assessment_datasets/student_performance.csv")

print("First 5 records:")
print(df.head())

print("\nLast 5 records:")
print(df.tail())

print("\nShape:")
print(df.shape)


print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nStatistical Summary:")
print(df.describe())
```

```python
# 12
print("\nFinal_Score is greater than 75:")

print("\nAttendance is below 75%:")
print(df[df['Attendance']< 75])

print("\nstudents from the Data Science department with Final_Score greater than 80:")
print(df[(df['Department'] == 'Data Science') & (df['Final_Score'] > 80)])

print("\n Final_Score from highest to lowest.:")
print( df.sort_values(by='Final_Score',ascending=False))
```

```python
# 13
score_columns = ['Python_Score','SQL_Score','Pandas_Score','NumPy_Score']

df['Average_Technical_Score'] = df[score_columns].mean(axis=1)

conditions = [
    df['Average_Technical_Score'] >= 90,
    df['Average_Technical_Score'] >= 75,
    df['Average_Technical_Score'] >= 60
]

choices = ['Excellent','Good','Average']

df['Performance'] = np.select(
    conditions,
    choices,
    default='Needs Improvement'
)

print(df)
```

```python
# 14
dept_summary = df.groupby('Department').agg(
    Number_of_Students=('Student_ID', 'count'),
    Average_Final_Score=('Final_Score', 'mean'),
    Maximum_Final_Score=('Final_Score', 'max'),
    Minimum_Final_Score=('Final_Score', 'min')
)

dept_summary_sorted = dept_summary.sort_values(by='Average_Final_Score', ascending=False)

print("GroupBy Analysis Summary (Sorted by Average Final Score descending):")
print(dept_summary_sorted)
```

```python
# 15
import numpy as np

scores_array = df['Final_Score'].to_numpy()

overall_mean = np.nanmean(scores_array)
overall_median = np.nanmedian(scores_array)
overall_std = np.nanstd(scores_array)
pct_25 = np.nanpercentile(scores_array, 25)
pct_75 = np.nanpercentile(scores_array, 75)

print("--- Final Score Statistics (using NumPy) ---")
print(f"Mean: {overall_mean:.2f}")
print(f"Median: {overall_median:.2f}")
print(f"Standard Deviation: {overall_std:.2f}")
print(f"25th Percentile: {pct_25:.2f}")
print(f"75th Percentile: {pct_75:.2f}")


students_above_mean = df[df['Final_Score'] > overall_mean]

print("\n--- Students with Final Score above the overall mean ---")
print(students_above_mean[['Student_ID', 'Name', 'Department', 'Final_Score']])
```

```python
# 16 a
print("Missing Values:")
print(df.isnull())

df.dropna(inplace=True)

print("\nAfter Handling Missing Values:")
print(df)

```

```python
# 16 b
import numpy as np

score_cols = ['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']

df['Average_Score'] = df[score_cols].mean(axis=1)

df['Pass_Status'] = np.where(df['Average_Score'] >= 50, 'Pass', 'Fail')

print("DataFrame with engineered features (first 10 rows):")
print(df[['Student_ID', 'Name', 'Average_Score', 'Pass_Status']].head(10))
```

```python
# 16 c
top_5_students = df.sort_values(by='Average_Score', ascending=False).head(5)
print("--- Top 5 Students based on Average_Score ---")
print(top_5_students[['Student_ID', 'Name', 'Average_Score']])

dept_avg_score = df.groupby('Department')['Average_Score'].mean()
highest_dept = dept_avg_score.idxmax()
highest_dept_score = dept_avg_score.max()
print(f"\n--- Department with Highest Average Score ---\n{highest_dept} ({highest_dept_score:.2f})")

avg_attendance = df.groupby('Department')['Attendance'].mean()
print("\n--- Average Attendance by Department ---")
print(avg_attendance)

pass_fail_counts = df['Pass_Status'].value_counts()
print("\n--- Number of Students Passed and Failed ---")
print(pass_fail_counts)
```

```python
#16 D
summary = df.groupby('Department').agg(
    Total_Students=('Student_ID', 'count'),
    Average_Score=('Average_Score', 'mean'),
    Average_Attendance=('Attendance', 'mean'),
    Pass_Count=('Pass_Status', lambda x: (x == 'Pass').sum()),
    Fail_Count=('Pass_Status', lambda x: (x == 'Fail').sum())
).reset_index()

print(summary.round(2))
```

```python
#17 A
sales_df = pd.read_csv("/content/drive/MyDrive/Data Science/Assessmment/numpy_pandas_assessment_datasets/sales.csv")


sales_df['Order_Date'] = pd.to_datetime(sales_df['Order_Date'], format='mixed', dayfirst=True)

print("Missing Values before handling:")
print(sales_df.isnull().sum())

sales_df['Sales'] = pd.to_numeric(sales_df['Sales'], errors='coerce')
sales_df['Discount'] = sales_df['Discount'].astype(str).str.replace('%', '', regex=False)
sales_df['Discount'] = pd.to_numeric(sales_df['Discount'], errors='coerce')

for col in ['Product', 'Category', 'Region']:
    if col in sales_df.columns:
        sales_df[col] = sales_df[col].astype(str).str.strip().str.title()

for col in ['Sales', 'Quantity', 'Discount', 'Profit']:
    if col in sales_df.columns:
        sales_df[col] = sales_df[col].fillna(sales_df[col].mean())

if 'Region' in sales_df.columns:
    sales_df['Region'] = sales_df['Region'].fillna(sales_df['Region'].mode()[0])

print("\nMissing Values after handling:")
print(sales_df.isnull().sum())
print("\nCleaned Sales DataFrame Preview:")
print(sales_df.head())
```

```python
#17 B
total_sales = sales_df['Sales'].sum()
total_profit = sales_df['Profit'].sum()
avg_sales = sales_df['Sales'].mean()

top_5_products = sales_df.groupby('Product')['Sales'].sum().reset_index().sort_values(by='Sales', ascending=False).head(5)

region_sales = sales_df.groupby('Region')['Sales'].sum().reset_index()
highest_sales_region = region_sales.sort_values(by='Sales', ascending=False).iloc[0]

print("--- Sales Analysis Summary ---")
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Average Sales per Transaction: ${avg_sales:,.2f}")

print("\n--- Top 5 Products based on Sales ---")
print(top_5_products.to_string(index=False))

print(f"\n--- Region Generating the Highest Sales ---\n{highest_sales_region['Region']} (${highest_sales_region['Sales']:,.2f})")
```
